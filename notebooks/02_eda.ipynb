```python
# Average goals scored by position

goals_by_position = (
    df.groupby("position")["goals_scored"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

goals_by_position.columns = ["position", "average_goals"]

goals_by_position
```

```python
# Average assists provided by position

assists_by_position = (
    df.groupby("position")["assists_provided"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

assists_by_position.columns = ["position", "average_assists"]

assists_by_position
```

```python
# Average dribbles per 90 by position

dribbles_by_position = (
    df.groupby("position")["dribbles_per_90"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

dribbles_by_position.columns = ["position", "average_dribbles_per_90"]

dribbles_by_position
```

```python
import matplotlib.pyplot as plt

# Average goals by position - Bar Chart

plt.figure(figsize=(8, 5))

plt.bar(
    goals_by_position["position"],
    goals_by_position["average_goals"]
)

plt.xlabel("Position")
plt.ylabel("Average Goals")
plt.title("Average Goals Scored by Position")

plt.show()
```
